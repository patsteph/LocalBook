# Accounts analysis

A quick look at the Harbor Lights accounts workbook.

```
import pandas as pd
df = pd.read_excel('accounts.xlsx', sheet_name='Accounts')
df.groupby('State')['ARR'].sum()
```

Texas carries the largest share of ARR. The Enterprise plan is a small share of accounts but a large share of revenue.
